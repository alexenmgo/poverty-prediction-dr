Serie de tiempo

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo en un dataframe de pandas


# Seleccionar las variables de interés y convertir la columna "ANO" a tipo datetime
data1['ANO'] = pd.to_datetime(data1['ANO'], format='%Y')

# Filtrar las columnas de pobreza, indigencia y año, excluyendo el año 2020
df_pobreza_indigencia = data1[data1['ANO'].dt.year != 2020][['ANO', 'pobre', 'indigencia']]

# Agrupar los datos por año y calcular el promedio de pobreza e indigencia
df_pobreza_indigencia = df_pobreza_indigencia.groupby('ANO').mean()

# Crear la serie de tiempo para la pobreza e indigencia
serie_pobreza = df_pobreza_indigencia['pobre']
serie_indigencia = df_pobreza_indigencia['indigencia']

# Configurar el gráfico de la serie de tiempo
plt.figure(figsize=(10, 6))
plt.plot(serie_pobreza.index.year, serie_pobreza, label='Pobreza')
plt.plot(serie_indigencia.index.year, serie_indigencia, label='Indigencia')
plt.xticks(serie_pobreza.index.year)
plt.xlabel('Año')
plt.ylabel('Porcentaje')
plt.title('Serie de Tiempo de Pobreza e Indigencia')
plt.legend()

# Agregar etiquetas a las líneas de tiempo
for x, y in zip(serie_pobreza.index.year, serie_pobreza):
    plt.text(x, y, f'{y:.2f}', ha='center', va='bottom')
for x, y in zip(serie_indigencia.index.year, serie_indigencia):
    plt.text(x, y, f'{y:.2f}', ha='center', va='bottom')

# Mostrar el gráfico
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_confusion_matrix(cm, classes, normalize=False, cmap='Blues'):
    """
    Función para visualizar una matriz de confusión con colores.

    Args:
        cm (array): Matriz de confusión.
        classes (list): Lista de nombres de las clases.
        normalize (bool, optional): Normalizar la matriz de confusión. Por defecto es False.
        cmap (str, optional): Colormap para los colores de la matriz. Por defecto es 'Blues'.
    """
    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

    fig, ax = plt.subplots()
    im = ax.imshow(cm, interpolation='nearest', cmap=cmap)
    ax.figure.colorbar(im, ax=ax)

    ax.set(xticks=np.arange(cm.shape[1]),
           yticks=np.arange(cm.shape[0]),
           xticklabels=classes, yticklabels=classes,
           title='Matriz de Confusión',
           xlabel='Clase Predicha',
           ylabel='Clase Real')

    thresh = cm.max() / 2
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(j, i, f"{cm[i, j]:.2f}",
                    ha="center", va="center",
                    color="white" if cm[i, j] > thresh else "black")

    fig.tight_layout()
    plt.show()

# Obtener la matriz de confusión
cm = confusion_matrix(y_test_p, y_pred_p_resampled)

# Definir los nombres de las clases
class_names = ['No pobre', 'Pobre']

# Visualizar la matriz de confusión con colores
plot_confusion_matrix(cm, class_names, normalize=False)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

# Define the models and their corresponding predictions
models = [
    ('Regresión Logística', y_pred_p_selected),
   # ('Árbol de Decisión', y_pred_pu),
    ('Random Forest', y_pred_p_res),
    ('SVM', y_pred_svc),
    ('XGBoost', y_predXGu),
    ('Redes Neuronales', y_predRN)
]

# Plot the ROC curve for each model
plt.figure(figsize=(8, 6))
for model_name, y_pred in models:
    fpr, tpr, _ = roc_curve(y_test_p, y_pred)
    roc_auc = auc(fpr, tpr)
    plt.plot(fpr, tpr, label=f'{model_name} (AUC = {roc_auc:.2f})')

# Set the labels and title for the plot
plt.xlabel('Tasa de falsos positivos')
plt.ylabel('Tasa de verdaderos positivos')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')

# Show the plot
plt.show()
```
